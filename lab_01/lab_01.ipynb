**№1 Створити 2 двовимірні масиви і виконати математичні операції над нимим**

```python
n = 3 # порядковий номер в журналі
matrix_size = n + 2

m1 = np.random.randint(0,10, size=(matrix_size, matrix_size)) # створення двовимірного масиву
m2 = np.random.randint(0,10, size=(matrix_size, matrix_size))

print("Матриця 1")
print(m1)
print("Матриця 2")
print(m2)
```

```python
sum = m1 + m2
print(sum)
```

```python
sub = m1 - m2
print(sub)
```

```python
div = m1 / m2
print(div)
```

```python
s_mul = m1 * 5 # скалярне множення
print(s_mul)
```

```python
v_mul = np.dot(m1,m2) # векторне множення
print(v_mul)
```

```python
v = np.linalg.det(m1) # визначник
print(v)
```

```python
d = np.linalg.eig(m1) # власні числа та вектори
print(d)
```

**№2 для заданої функції (1/ ln x x) визначити: похідну (в заданій точці), первісну (на заданому проміжку), а також виконати оптимізацію (визначити глобальний і локальний мінімуми).**

```python
import scipy
import scipy.integrate as integrate
import scipy.optimize as optimize
```

```python
def f(x):
    return 1 / (x * np.log(x)) # функція
```

```python
x0 = 2 # координати точки для визначення похідної
derivative = optimize.approx_fprime(x0, f, 1e-6) # визначення похідної 
print(f"Похідна у точці {x0}: {derivative[0]:.4f}")
```

```python
a,b = 1.5, 3.0 # проміжок для пошуку первсіної
integral_val, error = integrate.quad(f, a, b) # пошук первісної
print(f"Значення інтеграла на проміжку [{a}, {b}]: {integral_val:.4f}")
```

```python
local_min = optimize.minimize(f, x0=2.0)
print(f"\nЛокальний мінімум знайдено в точці x = {local_min.x[0]:.4f}")
print(f"Значення функції в цій точці: {local_min.fun:.4f}")
```

```python
bounds = [(1.5, 5.0)] 
global_min = optimize.differential_evolution(f, bounds)
print(f"\nГлобальний мінімум знайдено в точці x = {global_min.x[0]:.4f}")
print(f"Значення функції в цій точці: {global_min.fun:.4f}")
```
