## Завдання 1
**Освоїти основні команди візуалізації даних в бібліотеці Matplotlib**

```python
import pandas as pd
from numpy.random import exponential
import matplotlib.pyplot as plt # модуль pyplot дозволяє працювати з бібліотекою Matplotlib в оперативному стилі
```

```python
df = pd.DataFrame({'x':range(20), 'y':exponential(10,20)})
print(df)
```

```python
%matplotlib inline
```

```python
df.y.hist() # гістограму розподілу, де висота стовпчика показує кількість спостережень з таким значенням, яке показано по x
```

```python
data1 = exponential(5,20) # 20 спроб із середнім балом 5
data2 = exponential(6,20)
print(data1)
print(data2)
```

```python
plt.plot(data1); # графік успішності від номера спроби
```

```python
plt.scatter(range(len(data2)), data2,);
```

```python
plt.scatter(range(len(data2)), data2, color='red');
plt.title("Results", fontdict={'fontsize':20});
plt.xlabel("Attempt number");
plt.ylabel("Result");
```

```python
plt.plot(data1, color='green', marker='*', label="First player");
plt.scatter(range(len(data2)), data2, color='red', label="Second player");
plt.title("Results", fontdict={'fontsize':20});
plt.xlabel("Attempt number");
plt.ylabel("Result");
plt.grid(True);
plt.legend();
plt.xticks(range(0,20,2));
plt.savefig('results.png');
```

```python
# subplots треба щоб розмістити кілька графіків на одному спільному полотні
fig, axes = plt.subplots() # fig містисть усі зображення в яких 1 axes або більше; axes - полотно, яке може містити кілька графіків
axes.plot(exponential(5,20)); 
axes.set_title('Chart');
```

```python
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(10,5))
for row, row_axes in enumerate(axes): # Перебираємо кожен рядок сітки
    for column, ax in enumerate(row_axes): # Перебираємо кожeн графік рядка
        ax.plot(exponential(column, 20)); 
        ax.set_title('Canvas column {} row {}'.format(column+1, row+1));
        ax.grid(True);
fig.tight_layout();
```

```python
fig.savefig('all_results.png')
```

## Завдання 2
**Із датасетом виконати операції для роботи з візуалізації даних в Pandas**

```python
df = pd.read_csv('fifa_players.csv')
df.head()
```

```python
%matplotlib inline
```

```python
df.overall_rating.plot.hist(); # гістограма для конкретної колонки
```

```python
df.overall_rating.plot.hist(bins=20); # кількість колонок для групування даних
plt.show();
```

```python
df.overall_rating.plot.kde(); # густина розподілу
```

```python
df.columns
```

```python
df.plot.scatter(x='shot_power', y='weak_foot(1-5)');
```

```python
df.groupby('weak_foot(1-5)').shot_power.plot.kde();
plt.xlim(40,100); # діапазон значень
plt.legend();
```

```python
ax = df.age.plot.hist()
ax.set_title("Visualization");
ax.figure.savefig('something.pdf');
```

```python
fig, ax = plt.subplots(figsize=(10,5))
df.wage_euro.plot.kde(ax=ax, label="Зарплати гравців")
for label, reputation_df in df.groupby('international_reputation(1-5)'):
    print(label)
    reputation_df.wage_euro.plot.kde(ax=ax, label=label)
ax.set_xlim(0, 800000)
plt.legend()
plt.show()
```
